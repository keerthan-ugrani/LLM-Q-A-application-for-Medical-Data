```python
from langchain import PromptTemplate, LLMChain
from langchain.llms import GPT4All

# PATH = '/Users/henrismidt/Library/Application Support/nomic.ai/GPT4All/mistral-7b-openorca.gguf2.Q4_0.gguf'
PATH = (
    "C:/Users/luke/AppData/Local/nomic.ai/GPT4All/mistral-7b-openorca.gguf2.Q4_0.gguf"
)

llm = GPT4All(model=PATH, verbose=True)

prompt = PromptTemplate(
    input_variables=["question"],
    template="""
    Question: {question}
    
    Answer: Let's think step by step.
    """,
)

llm_chain = LLMChain(prompt=prompt, llm=llm)

question = "what are the most common forms of benign epilepsy syndromes for children?"
response = llm_chain.run(question)
response
```
