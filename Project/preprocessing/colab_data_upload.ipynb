```python
import os
from dotenv import load_dotenv, find_dotenv
from embedder import WeaviateCloudEmbedder

load_dotenv(find_dotenv("tokens.env"))
YOUR_WEAVIATE_KEY = os.getenv("YOUR_WEAVIATE_KEY")
YOUR_WEAVIATE_CLUSTER = os.getenv("YOUR_WEAVIATE_CLUSTER")
```

The Following Code takes approximatly 52min to execute:

```python
embedder = WeaviateCloudEmbedder(
    YOUR_WEAVIATE_KEY, YOUR_WEAVIATE_CLUSTER, device="cuda"
)
embedder.embed_data("data.json")
```

Test if similarity search works

```python
query = "children with benign childhood epilepsy"
docs = embedder.vectorstore.similarity_search_with_score(query, by_text=False)
print(docs[0])
```
