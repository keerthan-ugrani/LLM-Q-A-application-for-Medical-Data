```python
import weaviate

# import weaviate.classes as wvc
from dotenv import load_dotenv, find_dotenv
import os
from langchain.vectorstores.weaviate import Weaviate
from langchain.embeddings.huggingface import HuggingFaceEmbeddings


load_dotenv(find_dotenv("./env/tokens.env"))
WCS_API_KEY = os.getenv("YOUR_WEAVIATE_KEY")
WCS_CLUSTER_URL = os.getenv("YOUR_WEAVIATE_CLUSTER")

# client = weaviate.connect_to_wcs(
#     cluster_url=WCS_CLUSTER_URL,
#     auth_credentials=weaviate.auth.AuthApiKey(WCS_API_KEY),
#     # headers={
#     #     "X-OpenAI-Api-Key": os.environ["OPENAI_APIKEY"]  # Replace with your inference API key
#     # }
# )
# try:
#     # Wrap in try/finally to ensure client is closed gracefully
#     questions = client.collections.get("Question")

#     response = questions.query.near_text(
#         query="biology",
#         limit=2
#     )

#     print(response.objects[0].properties)  # Inspect the first object

# finally:
#     client.close()  # Close client gracefully

# For Version V4:
# client = weaviate.connect_to_wcs(
#     cluster_url=WCS_CLUSTER_URL,  # Replace with your WCS URL
#     auth_credentials=weaviate.auth.AuthApiKey(WCS_API_KEY)  # Replace with your WCS key
# )

# For Version V3:
client = weaviate.Client(
    url=WCS_CLUSTER_URL,
    auth_client_secret=weaviate.auth.AuthApiKey(WCS_API_KEY),
)

device = "cpu"
embed_model = HuggingFaceEmbeddings(
    model_name="BAAI/bge-small-en-v1.5",
    model_kwargs={"device": device},
    encode_kwargs={"device": device, "batch_size": 32},
)

# vectorstore = Weaviate(client, index_name="text", embedding=embed_model, text_key="text")
vectorstore = Weaviate(
    client,
    index_name="LangChain_0c70358e34034236ba8f84cd318e2c7b",
    embedding=embed_model,
    text_key="text",
    by_text=False,
)
```

```python
query = "children with benign childhood epilepsy"
docs = vectorstore.similarity_search_with_score(query)
print(docs[0])
```
