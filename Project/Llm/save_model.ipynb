```python
import weaviate

# import weaviate.classes as wvc
from dotenv import load_dotenv, find_dotenv
import os
from langchain.embeddings.huggingface import HuggingFaceEmbeddings
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline
import transformers
from torch import bfloat16


load_dotenv(find_dotenv("tokens.env"))
hf_auth = os.environ.get("HF_AUTH")
```

```python
model_id = "meta-llama/Llama-2-13b-chat-hf"

if torch.cuda.is_available():  # BitsAndBytesConfig needs a cuda GPU
    bits_and_bites_config = transformers.BitsAndBytesConfig(
        load_in_4bit=True,
        bnb_4bit_quant_type="nf4",
        bnb_4bit_use_double_quant=True,
        bnb_4bit_compute_dtype=bfloat16,
    )
    model = AutoModelForCausalLM.from_pretrained(
        model_id,
        trust_remote_code=True,
        quantization_config=bits_and_bites_config,
        device_map="auto",
        do_sample=True,
        token=hf_auth,
    )
else:
    model = AutoModelForCausalLM.from_pretrained(
        model_id,
        trust_remote_code=True,
        device_map="auto",
        do_sample=True,
        token=hf_auth,
    )
```

```python
model.save_pretrained("./Llm")
```
